## Exercício 1 – Separar colunas
Uma coluna é espectral se o cabeçalho pode ser convertido em número (`float("450")` funciona; `float("ID_Unico")` gera `ValueError`).

```python
def separar_colunas(df):
    meta, espec, wl = [], [], []
    for c in df.columns:
        try:
            wl.append(float(c))
            espec.append(c)
        except ValueError:
            meta.append(c)
    return meta, espec, np.array(wl)
```

## Exercício 2 – Ler com codificação automática
Tente [`utf-8`](../../praticas/curso_utils.qmd#cb1-38), depois [`latin1`](../../praticas/curso_utils.qmd#cb1-38), depois [`cp1252`](../../praticas/curso_utils.qmd#cb1-38). Devolva `(df, codificacao_usada)`.

```python
def ler_csv_auto(path):
    for enc in ('utf-8', 'latin1', 'cp1252'):
        try:
            return pd.read_csv(path, encoding=enc, low_memory=False), enc
        except UnicodeDecodeError:
            continue
    raise ValueError(f'Não foi possível ler {path}')
```

## Exercício 3 – Carregar NIR e MIR

```python
df_nir, enc = ler_csv_auto(DATA_DIR / NIR_FILE)
meta, espec, wl = separar_colunas(df_nir)
X_nir = df_nir[espec].apply(pd.to_numeric, errors='coerce').to_numpy(float)
print(X_nir.shape, wl.min(), wl.max(), enc)

df_mir, enc_mir = ler_csv_auto(DATA_DIR / MIR_FILE)
meta_mir, espec_mir, wl_mir = separar_colunas(df_mir)
X_mir = df_mir[espec_mir].apply(pd.to_numeric, errors='coerce').to_numpy(float)
print(X_mir.shape, wl_mir.min(), wl_mir.max(), enc_mir)
```
