## Exercício 2 – Checar dependências
Escreva [`checar_dependencias(pacotes)`](../../praticas/curso_utils.qmd#cb1-27): recebe [`{nome_pip: nome_modulo}`](../../praticas/curso_utils.qmd#cb1-27) e devolve a lista dos pacotes que **não** estão instalados. Use [`importlib.util.find_spec`](../../praticas/curso_utils.qmd#cb1-28).

```python
import importlib.util

def checar_dependencias(pacotes):
    return [pip for pip, mod in pacotes.items() if importlib.util.find_spec(mod) is None]

PACOTES = {'numpy': 'numpy', 'pandas': 'pandas', 'scikit-learn': 'sklearn', 'pyyaml': 'yaml', 'scipy': 'scipy'}
faltando = checar_dependencias(PACOTES)
print('Faltando:', faltando or 'nenhum')
```

## Exercício 3 – Ler parâmetros de um YAML
Leia [`Reference/parameters.yaml`](../../Trabalhos/01_pipeline_visualizacion_qc.ipynb#fundamentação-teórica) (se existir) e obtenha a janela do Savitzky-Golay com padrão 11. Se o arquivo não existir, use um dicionário vazio.

```python
import yaml

PARAMS_PATH = ROOT / 'Reference' / 'parameters.yaml'
if PARAMS_PATH.exists():
    with PARAMS_PATH.open(encoding='utf-8') as f:
        params = yaml.safe_load(f) or {}
else:
    params = {}

janela = params.get('preprocessing', {}).get('smoothing', {}).get('window_length', 11)
print('Janela SG:', janela)
```
